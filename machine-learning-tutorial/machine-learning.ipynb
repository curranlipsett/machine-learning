```python
# Course link on youtube https://www.youtube.com/watch?v=8jazNUpO3lQ&list=PLeo1K3hjS3uvCeTYTeyfe0-rN5r8zn9rw&index=3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("data/homeprices.csv")
df
```

```python
%matplotlib inline 
plt.xlabel("Area (sq ft)")
plt.ylabel("Price (US$)")
plt.scatter(df.area, df.price)
```

```python
rg = linear_model.LinearRegression()
# Build linear regression model by mappiny area to price of known data points
rg.fit(df[["area"]], df.price)
```

```python
rg.predict([[5000]])
```

```python
# Coefficient or slope of the linear regression line
rg.coef_
```

```python
# Intercept or B value in y = mx + B equation
rg.intercept_
```

```python
d = pd.read_csv("data/areas.csv")
d.head(3)
```

```python
p = rg.predict(d)
d["prices"] = p
d.to_csv("data/predicted_prices.csv", index=False)
```
