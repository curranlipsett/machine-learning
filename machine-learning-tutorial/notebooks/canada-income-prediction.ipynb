```python
# Course link on Github https://github.com/codebasics/py/tree/master/ML/1_linear_reg
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("data/canada_per_capita_income.csv")
df
```

```python
rg = linear_model.LinearRegression()
# Build linear regression model by mappiny area to price of known data points
rg.fit(df[["year"]], df.income)
```

```python
%matplotlib inline 
plt.xlabel("Year")
plt.ylabel("Income (CAD)")
plt.scatter(df.year, df.income, marker='x', color='purple')
plt.plot(df.year, rg.predict(df[['year']]), color='red', linestyle='dashdot')
```
