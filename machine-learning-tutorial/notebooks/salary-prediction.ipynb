```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("data/salary.csv")
df
```

```python
rg = linear_model.LinearRegression()
# Build linear regression model by mappiny area to price of known data points
rg.fit(df[["year"]], df.salary)
```

```python
%matplotlib inline 
plt.xlabel("Year")
plt.ylabel("Income (CAD)")
plt.scatter(df.year, df.salary, marker='x', color='purple')
plt.plot(df.year, rg.predict(df[['year']]), color='red', linestyle='dashdot')
```
